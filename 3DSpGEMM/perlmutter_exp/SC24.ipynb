## Load the log

```python
def read_csv(filename):
    data = pd.read_csv(filename,skiprows=1)
    data.drop(data.columns[len(data.columns)-1], axis=1, inplace=True) # remove last column 
    newindex = [x.strip() for x in data.columns ] # remove whitespace in the column title
    data.columns = newindex
    return data
```
