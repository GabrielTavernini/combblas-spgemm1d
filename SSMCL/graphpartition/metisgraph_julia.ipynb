```julia
str = "0 2 5 8 11 13 16 20 24 28 31 33 36 39 42 44"
jc =  split("0 2 5 8 11 13 16 20 24 28 31 33 36 39 42 44", " ")
jc = parse.(Int, jc)
ir =  split("1 5 0 2 6 1 3 7 2 4 8 3 9 0 6 10 1 5 7 11 2 6 8 12 3 7 9 13 4 8 14 5 11 6 10 12 7 11 13 8 12 14 9 13", " ")
ir = parse.(Int, ir)
println(jc)
println(ir)

row = Vector{Int64}()
col = Vector{Int64}()
nums = Vector{Int64}()
println("jc size ", length(jc))


for j in 1:length(jc)-1
    s = jc[j] + 1
    e = jc[j+1] + 1 - 1 # not include the end 
    for r in s:e
        push!(col,j)
        push!(row,ir[r] + 1)
        push!(nums,1)
    end
end

println(row)
println(col)
println(length(row))
println(length(ir))
sparr = sparse(row,col,nums,15,15)

mmwrite("myfile.mtx",sparr)
```
