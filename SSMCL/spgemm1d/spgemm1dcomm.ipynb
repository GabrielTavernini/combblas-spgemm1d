```python
def getprojarray(filename):
    M = int(filename[:-4].split("_")[1][1:])
    N = int(filename[:-4].split("_")[2][1:])
    K = int(filename[:-4].split("_")[3][1:])
    nprocs = int(filename[:-4].split("_")[4][6:])
    proj = np.fromfile(buildpath + "/" + filename,dtype=np.int32)
    Brows = proj.shape[0] // nprocs
    if Brows % nprocs != 0:
        print("error dimension!!")
    proj = proj.reshape((nprocs,Brows))
    print("M",M,"N",N,"nprocs",nprocs, "proj shape", proj.shape)
    return proj

filename = "BrowProj_M48_N48_K48_Nprocs16.bin"
proj = getprojarray(filename)
filenamecomm = "BrowProjCommInfo_M48_N48_K48_Nprocs16.bin"
projcomm = getprojarray(filenamecomm)

print ( np.allclose(proj, projcomm) )
```
